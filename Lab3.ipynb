```python
import heapq

# 1. Misplaced Tiles Heuristic

def misplaced_tiles(state, goal):
    count = 0

    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                count += 1

    return count


# 2. Manhattan Distance Heuristic

def manhattan_distance(state, goal):
    distance = 0

    for i in range(3):
        for j in range(3):

            if state[i][j] != 0:

                for x in range(3):
                    for y in range(3):

                        if state[i][j] == goal[x][y]:
                            distance += abs(i - x) + abs(j - y)

    return distance


# 3. Convert 2D puzzle into tuple

def convert(state):
    return tuple(tuple(row) for row in state)


# 4. Print puzzle

def print_puzzle(state):

    for row in state:
        print(row)

    print()


# 5. Generate neighboring states

def get_neighbors(state):

    neighbors = []

    # Find blank position
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                blank_i = i
                blank_j = j

    # Possible movements
    moves = [
        (-1, 0), # Up
        (1, 0), # Down
        (0, -1), # Left
        (0, 1) # Right
    ]

    for di, dj in moves:

        new_i = blank_i + di
        new_j = blank_j + dj

        # Check boundaries
        if 0 <= new_i < 3 and 0 <= new_j < 3:

            new_state = [list(row) for row in state]

            # Swap blank with neighboring tile
            new_state[blank_i][blank_j], new_state[new_i][new_j] = \
                new_state[new_i][new_j], new_state[blank_i][blank_j]

            neighbors.append(convert(new_state))

    return neighbors


# 6. A* Algorithm

def a_star(start, goal, heuristic):

    start = convert(start)
    goal = convert(goal)

    # Priority queue
    open_list = []

    # g(start) = 0
    g_cost = {start: 0}

    # f(start) = g(start) + h(start)
    f_cost = g_cost[start] + heuristic(start, goal)

    heapq.heappush(open_list, (f_cost, start))

    # Store parent of every state
    parent = {}

    # Already explored states
    closed = set()

    while open_list:

        # Get state with lowest f value
        _, current = heapq.heappop(open_list)

        # Goal reached
        if current == goal:

            path = []

            while current in parent:
                path.append(current)
                current = parent[current]

            path.append(start)

            path.reverse()

            return path

        # Skip if already explored
        if current in closed:
            continue

        closed.add(current)

        # Explore neighbors
        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            # Every move costs 1
            new_g = g_cost[current] + 1

            # If this is a better path
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                parent[neighbor] = current

                g_cost[neighbor] = new_g

                h = heuristic(neighbor, goal)

                f = new_g + h

                heapq.heappush(open_list, (f, neighbor))

    return None


# 7. Main Program

start = [
    [1, 2, 3],
    [4, 5, 6],
    [0, 7, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


# Using Misplaced Tiles

print("===== MISPLACED TILES =====")

path = a_star(start, goal, misplaced_tiles)

if path:

    print("Number of moves:", len(path) - 1)
    print()

    for state in path:
        print_puzzle(state)

else:
    print("No solution")


# Using Manhattan Distance

print("===== MANHATTAN DISTANCE =====")

path = a_star(start, goal, manhattan_distance)

if path:

    print("Number of moves:", len(path) - 1)
    print()

    for state in path:
        print_puzzle(state)

else:
    print("No solution")
```
